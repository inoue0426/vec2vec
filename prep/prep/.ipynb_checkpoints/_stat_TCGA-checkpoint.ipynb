```python
gex_features_df = pd.read_csv('code_ae_data/preprocessed_dat/uq1000_feature.csv', index_col=0)

with gzip.open('code_ae_data/raw_dat/Xena/TCGA_phenotype_denseDataOnlyDownload.tsv.gz') as f:
    xena_sample_info_df = pd.read_csv(f, sep='\t', index_col=0)

tcga_samples = xena_sample_info_df.index.intersection(gex_features_df.index)
tcga_unlabeled_df = gex_features_df.loc[tcga_samples]

print(tcga_unlabeled_df.shape)
tcga_unlabeled_df.head()
```
