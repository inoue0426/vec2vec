```python
import pandas as pd
import os
```

```python
folder = 'dataset/'
filenames = [f for f in os.listdir(folder) if 'tcga' in f and not f.endswith('.gz')]
filenames
```

```python
important_cols = [
    'project.project_id',
    'cases.submitter_id', 
    'diagnoses.primary_diagnosis',
    'treatments.treatment_outcome',
    'diagnoses.submitter_id',
    'treatments.submitter_id',
    'treatments.therapeutic_agents', 
]
```

```python
pd.read_table('dataset/clinical.project-tcga-gbm.2025-07-03/clinical.tsv')
```

```python
pd.read_table(file_path)[important_cols]
```

```python
res = pd.DataFrame()
summary_rows = []

for i in filenames:
    file_path = os.path.join('dataset/', i, 'clinical.tsv')
    try:
        df_all = pd.read_table(file_path)[important_cols]
        project_id = df_all['project.project_id'].iloc[0] if not df_all.empty else 'Unknown'
        total = len(df_all)

        # 薬が使われたレコードだけを抽出（therapeutic_agents が空でないもの）
        df_with_drug = df_all[df_all['treatments.therapeutic_agents'].notna() & 
                              (df_all['treatments.therapeutic_agents'] != "")]

        valid = len(df_with_drug)
        percent_valid = 100 * valid / total if total > 0 else 0

        summary_rows.append({
            'Project ID': project_id,
            'Total': total,
            'With Drug': valid,
            '%': round(percent_valid, 1)
        })

        res = pd.concat([res, df_with_drug], ignore_index=True)

    except Exception as e:
        summary_rows.append({
            'Project ID': f"{i} (Error)",
            'Total': 0,
            'With Drug': 0,
            '%': 0
        })

# 結果をDataFrameに
summary_df = pd.DataFrame(summary_rows)

# 合計行を追加
total_row = {
    'Project ID': 'Total',
    'Total': summary_df['Total'].sum(),
    'With Drug': summary_df['With Drug'].sum(),
    '%': round(100 * summary_df['With Drug'].sum() / summary_df['Total'].sum(), 1)
}
summary_df = pd.concat([summary_df, pd.DataFrame([total_row])], ignore_index=True)
```

```python
non_smiles_compounds = [
    'Not Reported',
    'Other',
    'Clinical Trial',
    'Clinical Trial Agent',
    'Hormone Therapy',
    'Total Androgen Blockade',
    'Dendritic Cell Vaccine',
    'PEP-3-KLH Conjugate Vaccine',
    'Antineoplastic Vaccine',
    'Multi-glioblastoma-peptide-targeting Autologous Dendritic Cell Vaccine ICT-107',
    'MAGE-A3 Peptide Vaccine',
    'AE37 Peptide/GM-CSF Vaccine',
    'Recombinant PRAME Protein Plus AS15 Adjuvant GSK2302025A',
    'Recombinant MAGE-3.1 Antigen',
    'Innate Immunostimulator rBBX-01',
    'Recombinant Adenovirus-p53 SCH-58500',
    'Light-Emitting Oncolytic Vaccinia Virus GL-ONC1'
]

res = res[~res['treatments.therapeutic_agents'].isin(non_smiles_compounds)]
res = res[res['treatments.treatment_outcome'] != 'Treatment Stopped Due to Toxicity']
res
```

```python
def recist_to_label(x):
    if x in ['Complete Response', 'Partial Response']:
        return 1
    elif x in ['Stable Disease', 'Progressive Disease', 'No Response']:
        return 0
    else:  # '--', 'Treatment Ongoing', 'Unknown', 'Not Reported'
        return 'Unknown'

res['recist_label'] = res['treatments.treatment_outcome'].apply(recist_to_label)

# 薬剤ごとにラベルごとの件数を集計
summary = res.groupby(['treatments.therapeutic_agents', 'recist_label']).size().reset_index(name='count')

# 見やすくピボット
pivot = summary.pivot(index='treatments.therapeutic_agents', columns='recist_label', values='count').fillna(0)
# pivot['total_known'] = pivot.get(0, 0) + pivot.get(1, 0)

# Total samples = 0+1+Unknown
pivot['total_samples'] = pivot.sum(axis=1)  # Unknown があれば自動で加算

# total_known で降順ソート
pivot_sorted = pivot.sort_values('total_samples', ascending=False)
```

```python
res = pd.DataFrame()
summary_rows = []

exclude_vals = ["--", "Unknown", "Not Reported", "Treatment Ongoing"]
valid_recist = ["Complete Response", "Partial Response", "Stable Disease", "Progressive Disease"]

for i in filenames:
    file_path = os.path.join('dataset/', i, 'clinical.tsv')
    try:
        df_all = pd.read_table(file_path)[important_cols]
        project_id = df_all['project.project_id'].iloc[0] if not df_all.empty else 'Unknown'
        total = len(df_all)

        df_valid = df_all[
            ~df_all['treatments.treatment_outcome'].isin(exclude_vals) &
             df_all['treatments.treatment_outcome'].isin(valid_recist)
        ]

        valid = len(df_valid)
        percent_valid = 100 * valid / total if total > 0 else 0

        summary_rows.append({
            'Project ID': project_id,
            'Total': total,
            'With RECIST': valid,
            '%': round(percent_valid, 1)
        })

        res = pd.concat([res, df_valid], ignore_index=True)

    except Exception as e:
        summary_rows.append({
            'Project ID': f"{i} (Error)",
            'Total': 0,
            'With RECIST': 0,
            '%': 0
        })

# 結果をDataFrameに
summary_df = pd.DataFrame(summary_rows)

# 合計行を追加
total_row = {
    'Project ID': 'Total',
    'Total': summary_df['Total'].sum(),
    'With RECIST': summary_df['With RECIST'].sum(),
    '%': round(100 * summary_df['With RECIST'].sum() / summary_df['Total'].sum(), 1)
}
summary_df = pd.concat([summary_df, pd.DataFrame([total_row])], ignore_index=True)
```

```python
summary_df.sort_values('With RECIST', ascending=False).reset_index(drop=True)
```
