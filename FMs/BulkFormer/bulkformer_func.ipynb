```python
import pandas as pd
import torch
```

```python
%load_ext autoreload
%autoreload 2
from BulkFormer import load_bulkformer, bulkformer_embed
```

```python
def choose_device():
    if torch.cuda.is_available():
        return "cuda"
    # if torch.backends.mps.is_available():
    #     return "mps"
    return "cpu"
```

```python
device = choose_device()
device
```

```python
log_tpm_df = pd.read_csv('data/demo.csv')
log_tpm_df.head()
```

```python
bf_obj = load_bulkformer(device=device)
```

```python
%%time
emb = bulkformer_embed(
    log_tpm_df=log_tpm_df,
    bf_obj=bf_obj,
    feature_type="transcriptome_level",
    aggregate_type="mean",   # "max" や "all" に変えてもOK
    batch_size=32,
    return_expr_value=False,
)

print(emb.shape)  # [N_samples, D]
```
